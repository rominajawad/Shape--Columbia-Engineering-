**Indentation:** Indentation (a series of space characters at the beginning of a line) is a syntactic element in Python!

* The `if` line together with an indented block of code is called a **compound statement**. 
* The `if` line is the **header** and the block is called the **suite** of the compound statement.

### Repetition: The `while` statement

* Repetition if a frequent component of algorithms.
* The `while` statement produces a *loop*. The indented block under the `while` header is run over and over until the boolean condition in the header becomes False.

`while ( boolean_expression ):  
    indented stuff here  
    more indented stuff here  
    this stuff is only executed if the boolean expression evaluates True
    the expression is checked again after each execution of the body
    if it's still true, we repeat execution of the loop body  `
    
    
A `while` loop is similar to an `if` statement except that it incorporates repition. When you want to perform
some task _until_ something happens, a while loop is the way to go.

```python
# Examples: Compute the base 2 logarithm of a positive integer number. 
# Assume we have a power of 2. 

n = 5
original_n = n
counter = 0

while n > 1:    
    n = n // 2
    counter += 1 # counter = counter + 1 #increment counter
    
print("log_2 of",original_n, "is", counter)
    

```

```python
# Collatz Conjecture a.k.a 3N+1 conjecture 
# Start at any positive integer n 
# If n is even: divide n by 2
# If n is odd: set n = 3*n + 1
# Conjecture: after a finite number of steps you reach 1

n = 13
print(n)

counter = 0
while n != 1:  # not n == 1 
    if n % 2 == 0: # modulo returns remainder of integer division
        n = n // 2
        
    else: 
        n = 3 * n + 1
    print(n)
    counter += 1

print("Done. Total steps: ", counter)
        
```

### The `break` statement
* Sometimes it is convenient to interrupt a `while` loop before we finish the suite of the statement.
* `break` immediately terminates the inner-most loop.
* Warning: some people consider this bad style (you can no longer predict the looping behavior by looking at the variables in the boolean expression only)

The `break` statement is often used when your computation produced an answer and there is no point to continue further. 

In our Rock/Paper/Scissors game, we can use `break` to merge the "play again" question into the move selection.

### `continue` statement
The `continue` statement skips the remaining lines of the suite and jumps back to the header. 

In the Rock/Paper/Scissors game we can use `continue` to simplify the input verification.

```python
x = 0 
while x < 10:
    x += 1
    if x % 2 == 1:
        continue 
    print(x)
    
```

### Iteration: The `for` statement
* A common operation is to perform some part of an algorithm on each element of a *collection* of items.
* A **collection** is a single object in Python that contains multiple elements, for example a list [0,5,8,9] or a string "hello".
* Most collections in Python are also **iterables** which means that the `for` statement can iterate over them.
* We will see how to deal with lists and other collections later in this course.

`for item_variable in iterable:  
    repeat some set of indented statements  
    where with each iteration, the variable    
    item_variable is associated with he next element in the list `


```python
li = ['some','stuff', -42, True, None]

i = 0 
for e in li: 
    print(e, "at index", i)
    i += 1
    
```

All for loops can be rewritten using while loops

```python
i = 0 

while i < len(li):
    print(li[i],"at index", i)
    i += 1
```

The built-in `range` function can be used to create a sequence of numbers to iterate over.

```python
result = []

for x in range(3,13):
    if x % 2 != 0:        
        result.append(x)

print(result)
```

Sometimes we need to generate combinations of items from multiple lists. We can use nested `for` loops. 

```python
colors = ['navy','white','red','peach','octarine']
clothes = ['pantsuit','shirt','dress','jacket','wizard robe']
for x in colors: 
    for item in clothes: 
        print(x, item)
```
