### Nonprinting Characters

Nonprinting Characters are special characters that *control* the output, rather than printing anything. 

The two you should remember are
* carriage return \n, which ends the current line and starts a new one
* tab \t, which inserts a tab character and indents the line.

````python
x = "first\nsecond"
x
````

````python
print(x)
````

````python
print("first\n\tsecond\tthird\n\tfourth\t\tfifth")
````

### String Encodings

Strings are just sequences of characters. Each character is internally encoded and stored as a number. When you print a string, how does the interpreter know which character to display?

There is a table of characters called an encoding. The specific encoding Python 3 uses by default is UTF-8. UTF-8 is a Unicode encoding.

ASCII is a subset of UTF-8.


````python
ord('a') # you can actually find the numeric value of a character
````

````python
chr(97) # You can also create a character from its numeric value
````

### String Formatting

Often we want to use a string to print some information in user-friendly format.
We can define a string containing place-holders and then fill in these place-holders using the `format` method. 


````python
x = "{}'s age is {}"
````

````python
print(x)
````

````python
x.format('Anna',28)
````

````python
name = "Anna"
age = 28
````

````python
x.format(name, age)
````

````python
name + " is "+ str(age) + " years old"
````

````python
"{}'s favorite color is {}".format('Galahad','YELLOWWWWW')
````

The {} place holders can contain additional information about how to format each argument.
For example, we often want to control the precision of floating point numbers.

````python
import math
math.pi
````

````python
"Pi is {}".format(math.pi)
````

````python
print("Pi is {:.2f}".format(math.pi))
````

````python
print("Pi is " + str(math.pi)[:4])
````
