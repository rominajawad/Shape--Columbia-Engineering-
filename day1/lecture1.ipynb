## Representing Data

* Computers use *switches* to represent information.
    * A switch is either on or off (electricity flows or not)*
    * Transistors are commonly used in computers as switches which can be either in an ”on” or ”off” state.*
    * Transistors are smaller and smaller and smaller* 
    * Microprocessor chips are a bunch of transistors*
    
    
### Binary Representation:     
*  The switch nature of transistors make storing numbers in binary a natural fit.
   * Binary is a change of base for our number system. 
   * In a number, its position represents powers of 2.     
   * Decimal uses digits 0-9 and positions in a number as powers of 10
   * Binary uses digits 0,1 and positions in a number as powers of 2 
   * We can convert back and forth (more later)

### Units of Data Storage
* Bit: binary digit, a single 1 or 0
* Byte: 8 bits e.g., e.g. 10110010
* Word: A computer’s standard unit of storage. A typical computer today is a 64 bit computer, meaning its word size is 64 bits (8 bytes). 

We use Byte to talk about storage. E.g., 2GB memory. Storage devices come in large quantities.
* 1K (kilobyte, kB)= 10^3 bytes, is about 2^10 bytes = 1024 bytes (kibibyte)
* 1M (megabyte, MB)= 10^6 bytes, is about 2^20 bytes = 1,048,576 bytes (mebibyte)
* 1G (gigabyte, GB)= 10^9 bytes, is about 2^30 bytes = 1,073,741,824 bytes (gibibyte)
* 1T (terabyte, TB)= 10^12 bytes, is about 2^40 bytes = 1,099,511,627,776 bytes (tebibyte)

Rule of thumb: Every 10 powers of 2 is about 3 powers of 10.

#### Size Comparisons

* 100kB = a typewritten page

### Storing Text and Other Data

Text = Strings are just sequences of characters. Each character is internally encoded and stored as a number. 

There is a table of characters called an **encoding.** 

* For English, the traditional encoding is ASCII (American Standard Code for Information Interchange). 
* **The Unicode Standard** is a global standard for encodings, that aims to represent all human writing systems. 
* UTF-8 is a common implementation of Unicode, and a superset of ASCII. It uses up to 4 bytes per character.


* Data is stored in files -- there are two types: 
    * Text files (each character in some encoding represents a letter in a word), human readable in any text editor. Computer program source code (e.g. in Python) is normally stored as textfiles. 
    * Binary files, typically used for multimedia, cannot be read without special software.  
    * Data compression reduces the size required to store data algorithmically.
