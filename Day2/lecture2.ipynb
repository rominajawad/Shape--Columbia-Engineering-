### How should you name variables?
* Python has some rules about names: 
    * A name may be of arbitrary length and may contain letters, digits, and underscores (\_) .
    * Every name must begin with a letter or underscore.
    * A name must NOT be a *keyword* (e.g. `True`)
    * Names are case sensitive.
* It's important to use good variable names for readability!
    * The name should tell you you something about what data the variable is associated with or what it is used for. 
    * For variable names, stick with lower-case letters and \_.
    * There are style guides (will distribute one next week).
    
**Rule 4: A foolish consistency is the hobgoblin of little minds.**
