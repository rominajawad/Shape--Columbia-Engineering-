### Graphical User Interfaces
* A graphical user interface (GUI) allows to interact with a computer program.
* The interaction is made through a pointing device such as a mouse, a stylus pen, or a human finger on a touch screen.
* A GUI program’s interface is composed of widgets displayed in a window.
* The pointing device can manipulate the widgets on a computer screen.
* Question: How are GUI programs different from Command Line Interface python programs?

### Event Driven Programs

* In an event driven program, program operations are initiated by certain events (user input, timers, ...) 
* When an event occurs, we say the event is **fired**.
* We need to write **event handlers** for these events.
* In a GUI, events often involve clicking or dragging. We need to **bind** event handlers to GUI elements such as buttons etc. 



### Building a GUI

1. Create the widgets that are displayed to a user and organize
them inside a screen window.
2. Define functions that will process user and application
events.
3. Associate specific user events (clicking a button etc.) with specific functions.
4. Start an infinite **event-loop** that processes user events.
When a user event happens, the event-loop calls the function
associated with that event.

### TkInter
There are many different GUI toolkits. A popular one for Python is TkInter ("TK interface").

Tk was originally developed for another programming language, TCL, and TkInter just adds a "thin object oriented layer" on top of Tk. 

TkInter comes with a number of "Widgets" that are defined as Python classes. Each instance of these classes corresponds to a GUI element. 

* **Tk** The top level window. 
* **Button** A simple button, used to execute a command or other operation.
* **Canvas** This widget can be used to draw graphs and plots, create graphic editors, and to implement custom widgets.
* **Checkbutton** Represents a variable that can have two distinct values. Clicking the button toggles between the values.
* **Entry** A text entry field.
* **Frame** A container widget. The frame can have a border and a background, and is used to group other widgets when creating an application or dialog layout.
* **Label** Displays a text or an image.
* **Radiobutton**  Represents one value of a variable that can have one of many values. Clicking the button sets the variable to that value, and clears all other radiobuttons associated with the same variable.
* **Scrollbar** Standard scrollbars for use with canvas, entry, and text widgets.
* **Text** Formatted text display. Allows you to display and edit text with various styles and attributes. Also supports embedded images and windows.
* ...

 ### Nested Functions
Typically, our GUI application will need access to the widgets, as well as other information stored as part of the program. 

```python
from tkinter import Tk, Label, Button

def run_gui():    


    root = Tk()
    root.wm_title("First test window")

    count = [0]

    label = Label(master=root, text=str(count[0]), font=("Times",20)) # need to specify the parent
    label.pack(side="top") # add the label into the root window

    
    def button_clicked():
        count[0] = count[0] + 1
        label['text'] = str(count[0])

    button = Button(root, text="Increment", command=button_clicked)
    button.pack(side="top")

    root.mainloop()
```

```python
run_gui()
```

### Using an object to store the data

```python
class GUIApp(): 
    
    def __init__(self):
        self.root = Tk()
        self.root.wm_title("First test window")
        self.count = 0

        self.label = Label(master=self.root, text=str(self.count), font=("Times",20)) # need to specify the parent
        self.label.pack(side="top") # add the label into the root window
        
        self.button = Button(master=self.root, text="Increment", command=self.button_clicked)
        self.button.pack(side="top")
        self.root.mainloop()

    def button_clicked(self):
        self.count += 1
        self.label['text'] = self.count 
```

Let's also set the size of this window. An easy way is to use a Frame, which groups together widgets. Frames can be given a size. 

```python
from tkinter import Frame

class GUIApp(): 
    
    def __init__(self):
        self.root = Tk()
        self.root.wm_title("First test window")
        self.count = 0

        self.frame = Frame(master=self.root, height=400, width=600)
        self.frame.pack_propagate(0) # don't shrink
        
        self.label = Label(master=self.frame, text=str(self.count), font=("Times",20)) # need to specify the parent
        self.label.pack(side="top") # add the label into the root window
        
        self.button = Button(master=self.frame, text="Increment", command=self.button_clicked)
        self.button.pack(side="top")
        
        
        self.frame.pack(side="top")
        
        self.root.mainloop()

    def button_clicked(self):
        self.count += 1
        self.label['text'] = self.count 
        
app = GUIApp()
```

Let's add a checkbox

```python
from tkinter import Checkbutton, IntVar

class GUIApp(): 
    
    def __init__(self):
        self.root = Tk()
        self.root.wm_title("First test window")
        self.count = 0

        self.frame = Frame(master=self.root, height=400, width=600)
        self.frame.pack_propagate(0) # don't shrink
        
        self.label = Label(master=self.frame, text=str(self.count), font=("Times",20)) # need to specify the parent
        self.label.pack(side="top") # add the label into the root window
        
        self.button = Button(master=self.frame, text="Increment", command=self.button_clicked)
        self.button.pack(side="left")
        
        self.exit_button = Button(master=self.frame, text="Exit", command=self.root.destroy)
        self.exit_button.pack(side="right")
        
        
        self.checkbox_var = IntVar()
        self.checkbox_var.set(1)
        
        self.checkbox = Checkbutton(master=self.frame, text="subtract", var = self.checkbox_var,
                                   onvalue = -1, offvalue = 1)
        self.checkbox.pack()
        
        self.frame.pack(side="top")
        
        self.root.mainloop()

    def button_clicked(self):
        self.count += self.checkbox_var.get()
        self.label['text'] = self.count 
        
app = GUIApp()
```

In order to use the checkbox, we need to tie it to a **TkInter variable**. These are a little counterintuitive. Basically, think of an object that stores a single value. This value is set when the user clicks the checkbox and it can then be queried. 

We can also use radio buttons. These are created by specifying a Radiobutton instance for each option. Options are grouped together because they share the same variable. 

```python
from tkinter import Radiobutton 

class GUIApp(): 
    
    def __init__(self):
        self.root = Tk()
        self.root.wm_title("First test window")
        self.count = 0

        self.frame = Frame(master=self.root, height=400, width=600)
        self.frame.pack_propagate(0) # don't shrink
        
        self.label = Label(master=self.frame, text=str(self.count), font=("Times",20)) # need to specify the parent
        self.label.pack(side="top") # add the label into the root window
        
        self.button = Button(master=self.frame, text="Increment", command=self.button_clicked)
        self.button.pack(side="left")
        
        self.exit_button = Button(master=self.frame, text="Exit", command=self.root.destroy)
        self.exit_button.pack(side="right")
        
        
        self.checkbox_var = IntVar()
        self.checkbox_var.set(1)
        
        self.checkbox = Checkbutton(master=self.frame, text="subtract", var = self.checkbox_var,
                                   onvalue = -1, offvalue = 1)
        self.checkbox.pack()
        
        self.radio_var = IntVar()
        self.radio_var.set(10)
        self.option1 = Radiobutton(master=self.frame, var = self.radio_var, text="1", value=1)
        self.option2 = Radiobutton(master=self.frame, var = self.radio_var, text="10", value=10)
        self.option3 = Radiobutton(master=self.frame, var = self.radio_var, text="100", value=100)
        self.option1.pack(side="left")
        self.option2.pack(side="left")
        self.option3.pack(side="left")
        
        
        self.frame.pack(side="top")
        
        self.root.mainloop()

    def button_clicked(self):
        self.count += self.checkbox_var.get() * self.radio_var.get()
        self.label['text'] = self.count 
        
app = GUIApp()
```

Finally, you can also use event-driven programming to set up timers and run functions once the timer expires. This is useful in games. 

```python
from datetime import datetime
str(datetime.now())
```

```python
class GUIApp(): 
    
    def __init__(self):
        self.root = Tk()
        self.root.wm_title("First test window")

        self.frame = Frame(master=self.root)
        #self.frame.pack_propagate(0) # don't shrink
        
        self.label = Label(master=self.frame, text=str(datetime.now()), font=("Arial",40)) # need to specify the parent
        self.label.pack(side="top") # add the label into the root window

        self.frame.pack(side="top")
        
        self.root.after(1000,self.update_clock)
        self.root.mainloop()

    def update_clock(self):
        self.label['text'] = str(datetime.now())
        self.root.after(1000,self.update_clock)
    
        
        
app = GUIApp()
```

We can now add shapes to the canvas. 

```python
from tkinter import Tk, Label, Button, Frame, Canvas
from datetime import datetime


def run_gui():    
    
    root = Tk()
    root.wm_title("First test window")

    frame = Frame(master=root, height=400,width=400)
    frame.pack_propagate(0) # don't shrink
    
    canvas = Canvas(master=frame, height=300, width=300, bg="gray")
    
    canvas.create_rectangle(0,0, 100,100, fill="blue")
    canvas.create_line(0, 50, 200, 0, fill="green", width=3)
        
    canvas.create_text(150,150,text="test",font=("Arial",20))
            
    canvas.create_oval(100,20,200,50, fill="red")    
            
    canvas.pack()
    frame.pack()
    root.mainloop()         
    
    
run_gui()
        
```

We would also like to add some interactivity. This can be done by registering certain events on the canvas instance and connecting these events to callback functions. 

As before, we will use a class to represent the application and then add the event handler as a method. 

```python
from tkinter import Tk, Label, Button, Frame, Canvas
from datetime import datetime



class CircleApp():
    def __init__(self):    

        self.root = Tk()
        self.root.wm_title("Circle application")

        self.frame = Frame(master=self.root, height=400,width=400)
        self.frame.pack_propagate(0) # don't shrink

        self.canvas = Canvas(master=self.frame, height=300, width=300, bg="white")

        self.canvas.bind("<Button-1>", self.handle_click)

        self.canvas.create_rectangle(0,0, 100,100, fill="blue")

        
        self.canvas.pack()
        self.frame.pack()
        self.root.mainloop()       
                           
    def handle_click(self, event):
        self.canvas.delete("all")
        if event.x > 100: 
            self.canvas.create_oval(event.x-10,event.y-10,event.x+10,event.y+10, fill="red")            
        else: 
            self.canvas.create_oval(event.x-10,event.y-10,event.x+10,event.y+10, fill="green")    
    
    
app = CircleApp()        
```
