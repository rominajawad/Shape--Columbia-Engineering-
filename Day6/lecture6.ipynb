### Data Structures
A **Data Structure** is an organizational scheme for a collection data, together with a specification of the operations that can be performed on the collection. 

We have already discussed **Data Types**, which are types of objects in Python that support certain operations. The concept of data structures is similar, except that it focuses on collections of data, rather than of a single concrete object. 

When designing data structures, the focus is usually on efficiency. Data structures are designed to support efficient algorithms, so the data structure operations should be as efficient as possible. 

Example: Imagine an algorithm that needs to search repeatedly for an item in a collection. How could you store the data more efficiently? 

Data structure implementations vary in their time and space efficiency. There is usually a trade-off between time and space requirements. 

Examples for data structures: 
* Lists, stacks, queues, dictionaries/maps (implemented using binary search trees or hash tables), sets, graphs, matrices, ...

## Dictionaries                 

Another important data structure is the **dictionary** (sometimes called **maps** or **associative arrays**).

Dictionaries are collections of items, but unlike lists or tuples, they are not sequences.

Instead, think of a dictionary as a collection of (*key*:*value*) pairs. 
  * The keys in the dictionary are unique (there is a *set* of keys), but the values may not be. 
  * Dictionaries make the lookup (get) operation for keys (and thus for values) very efficient.
  * In the Python dictionary, this efficient lookup is accomplished through *hashing*. As a result, keys need to be immutable. 

Dictionaries are the most powerful built-in data structure.

We will discuss hashing and hash tables in a later segment.

### Some dictionary operations

```python
my_dict = {'a':1, 'answer':42, 3:{1:'x',2:'y'}}
```

```python
len(my_dict) # number of key:value pairs
```

```python
li = ['a','b','c','d']
'd' in li
```

As for lists, you can use a boolean expression with `in` to check for membership. This is a lot more efficient with a dictionary.  

```python
'a' in my_dict # boolean expression to check for membership.
```

```python
'c' in my_dict
```

```python
1 in my_dict
```

You can use a `for` loop to iterate through the keys of the dictionary. Note that you cannot expect these to be in any particular order!

```python
for x in li: 
    print(x)
```

```python
for key in my_dict:
    print(key)
```

How would you iterate through the values of the dictionary?

```python
for x in my_dict: 
    value = my_dict[x]
    print(x, '-->', value)
```

```python
for value in my_dict.values():
    print(value)
```
