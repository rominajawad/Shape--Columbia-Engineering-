##  Session 7: Introduction to Functions 



* Functions in Python work similar to functions in mathematics y = f(x).
* A function perform some operation given one or more parameters and return one value.

* Functions ”encapsulate” the performance of some particular operation, so it can be reused by others (for example, the sqrt() function)

for example: f(x) => √x

````python
import math
y = math.sqrt(10)
y
````

Functions are a way to solve a specific sub-problem. They **abstract** away from the actual solution to that subproblem.

You do not have to know *how* the math.sqrt function is implement, just what it computes. 

This encourages top-down problem solving:  
   * divide the problem into sub-problems.
   * solve each individual sub-problem (write a function fo reach), then combine the solutions.
   * you can **reuse** existing functions you or someone else wrote earlier and **share** your functions with other programmers. 
   * you can even design an algorithm assuming you have a function that solves the subproblem, and then implement the function later (onion approach). 
   
Functions can also make your programs **easier to read and maintain**:
  * Its easier to debug a single function than a complete program. 
  * You avoid redundancy, because you implement functionality only once. If it breaks, you only need to fix one function. 
  * A single function is also easier to make secure.
  
As we will see later, functions are also a powerful computational device. They allow **recursion**.  

### Defining and Invoking Functions

Functions are defined using the `def` keyword. 

```def function_name ( parameter_1 , ... , parameter_n ) :
    """
    A docstring describing the function .
    """
    statements
    ...
    return result```
    
A function **definition** consists of
   * a function **name** (must follow variable naming rules)
   * zero or more comma separated parameters **parameters**.
   * a function **body** or **suit** (a list of indented statements)
   * one or more **return statements** 

````python
def f(celsius_float): 
    return celsius_float * 1.8 + 32
````

````python
f(100)
````

Once a function is defined, it can be **invoked** (or **called**). A function must be defined *before* it is invoked. 
 * When a function is invoked, **arguments** are passed into the function's parameters. 
 * A function invokation is an expression whose value is the return value of the function. 

````python
x = f(25.0) # 25 is the argument passed into the celsius_float parameter
x
````

````python
def fact(n): # another example
    """
    Compute n! = 1 * 2 * 3 * 4 * 5 * ... * n 
    """
    total = 1
    for i in range(1,n+1):
        total = total * i 
        
    return total 
````

````python
fact(3)
````

`return` can occur anywhere in the function (not necessarily the last line). 
   * `return` terminates the function and returns the return value. 
   * a `return` statement without a return value returns `None`
   * A function with no return statement returns `None` once  there are no more statements to execute.
   * There can be multiple return statements. 
   
Every function has a **return value**, even if there is no return statement.   

````python
def pos_only(x): #aka the rectifier function 
    if x > 0:
        return x #terminates the function right away
     
    return 0
````

````python
pos_only(-5)
````

Functions can have multiple parameters

````python
import math

def pythagoras(leg_a, leg_b):
    """
    Compute  the  length  of the  hypotenuse
    opposite  of the  right  angle  between leg_a
    and leg_b.
    """
    hypotenuse = math.sqrt(leg_a **2 + leg_b **2)
    return  hypotenuse
````

````python
pythagoras(3.0, 4.0)
````

### Function Control Flow 

* When a function is called, the main program (or another function) is temporarily suspended. 
* Control is then passed to the function. 
* Once the function finishes and returns the return value, the main program continues.
* You can "nest" arbitrarily many functions this way. This mechanism enables *recursion*.

### Name space of a function

Every function defines its own namespace. 
* Variables defined inside a function are not "visible" outside the function.

````python
x = "hi"

def foo(x): # This x is entirely different!!! 
    print(x)  # x shadows the x in the surrounding namespace
    return x
    
foo(27)   
print(x)
````

* However, variables defined outside the function are visible inside the function.

````python
some_int = 42
def foo():  
    print(some_int)  # Closure -- some_int defined in the context of the function
    return None
````

````python
foo()
````

````python
some_int = 23
some_int
````

````python
foo()
````

* The value of these variables is the value at the time when the function is *called*, not the value it has when the function is defined. 

*Careful:* When, within a function, you assign something to a variable name that exists in the surrounding namespace, you define a *new* local variable.  

````python
some_int = 23
def foo():  
    #some_int inside of foo shadows some_int in the surrounding namespace 
    print(some_int)
    some_int = 42 # this is a new variable in foo's namespace, that happen to have the same name as some_int

foo()
print(some_int)    
````

### Program Development: When to use a Function

How do you know when to write a function to do something?

* A function should **do one thing**: a clearly identifyable operation (ideally one that you would use more than once). 
* A function should be **readable** (and **not too long**: most functions aren't longer than a few lines or at most a page of code).
* A function should be **reusable**. It should not depend on the context in which it is being defined, but should be **self-contained**.
* Think about all possible input cases when defining a function. The function should be **complete**. 
* **refactoring** is the process of restrucuring your code to improve it, without changing functionality. One example is to split parts of your code off into functions. 


** Calling functions from the main program **

````python
def celsius_to_fahrenheit(celsius):
    return celsius * 1.8 + 32

celsius_t = float(input("Enter the temperature in celsius: "))
fahrenheit_t = celsius_to_fahrenheit(celsius_t)
print("That's {:.2f} in Fahrenheit.".format(fahrenheit_t))
````

When you write programs, a good practice is to wrap the main program in the following: 

```if __name__ == "__main__":

This way, when you import the file as a module, the program won't run, but you will still be able to call any functions from another program. 

A **module** is like a library of functions (and other things) that you can access. We have already seen some modules that come built into Python. 

To write your own, just put your functions into a file named module_name.py, then import them from another program with 

`import module_Name`

**Recursion teaser:**

````python
def fact(n):  # n! = 1 * 2 * ... * n = n * (n-1) * (n-2) * ... *1
             #    = n * (n-1)!
    if n == 1: 
        return 1
    return n * fact(n-1)
````

````python
fact(5)
````
